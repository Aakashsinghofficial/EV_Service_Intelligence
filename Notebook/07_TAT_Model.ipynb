# Turnaround Time Prediction

## Objective

Predict the actual turnaround time of an EV service visit using
information available at the prediction point.

### Target
Turnaround_Time_Days

### Problem Type
Regression

### Models
- Linear Regression
- Decision Tree Regressor
- Random Forest Regressor

### Evaluation Metrics
- MAE
- RMSE
- R²

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
df = pd.read_csv("ml_ready.csv")
```

```python
df.shape
```

```python
X = df.drop(
    columns=[
        "Repair_Cost",
        "Turnaround_Time_Days",
        "Is_Delayed"
    ]
)

y = df["Turnaround_Time_Days"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "bool"]
).columns.tolist()
```

```python
print("Numerical Features:", len(numerical_features))
print(numerical_features)

print("\nCategorical Features:", len(categorical_features))
print(categorical_features)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            "passthrough",
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)
            
```

```python
linear_model = Pipeline(
    steps = [
        ("preprocessor",preprocessor),
        ("model",LinearRegression())
    ]
)
```

```python
linear_model.fit(X_train,y_train)
```

```python
y_pred_linear = linear_model.predict(X_test)
```

```python
mae_linear = mean_absolute_error(
    y_test,
    y_pred_linear
)

rmse_linear = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_linear
    )
)

r2_linear = r2_score(
    y_test,
    y_pred_linear
)
```

```python
print(f"MAE  : {mae_linear:.2f}")
print(f"RMSE : {rmse_linear:.2f}")
print(f"R²   : {r2_linear:.3f}")
```

```python
tree_model = Pipeline(
    steps=[
        ("preprocessor",preprocessor),
        (
            "model",
            DecisionTreeRegressor(
                random_state=42
            )
        )
    ]
)
```

```python
tree_model.fit(X_train,y_train)
```

```python
y_pred_tree = tree_model.predict(X_test)
```

```python
mae_tree = mean_absolute_error(
    y_test,
    y_pred_tree
)

rmse_tree = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_tree
    )
)

r2_tree = r2_score(
    y_test,
    y_pred_tree
)

print(f"MAE  : {mae_tree:.2f}")
print(f"RMSE : {rmse_tree:.2f}")
print(f"R²   : {r2_tree:.3f}")
```

```python
forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
forest_model.fit(X_train, y_train)
```

```python
y_pred_forest = forest_model.predict(X_test)
```

```python
mae_forest = mean_absolute_error(
    y_test,
    y_pred_forest
)

rmse_forest = np.sqrt(
    mean_squared_error(
        y_test,
        y_pred_forest
    )
)

r2_forest = r2_score(
    y_test,
    y_pred_forest
)

print(f"MAE  : {mae_forest:.2f}")
print(f"RMSE : {rmse_forest:.2f}")
print(f"R²   : {r2_forest:.3f}")
```

```python
results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest"
    ],
    "MAE": [
        mae_linear,
        mae_tree,
        mae_forest
    ],
    "RMSE": [
        rmse_linear,
        rmse_tree,
        rmse_forest
    ],
    "R2": [
        r2_linear,
        r2_tree,
        r2_forest
    ]
})

results.sort_values("MAE")
```
